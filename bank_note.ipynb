Identifikasi dataset:

Jenis Klasifikasi: Binary Classification.
- Independent Variables: Variance, Skewness, Curtosis, dan Entropy.
- Dependent Variable: class.
- Jumlah Independent Variables: 4 fitur.
- Jumlah Dependent Variables: 1 target.
- Jumlah Kategori Target: 2 kategori.
- Kategori 0: Uang asli.
- Kategori 1: Uang palsu.

```python
import pandas as pd

df = pd.read_csv("BankNote_Authentication.csv")

print("IDENTIFIKASI DATASET BANK NOTE")

print("\nInformasi Dataset:")
df.info()

print("\nJumlah Baris dan Kolom:")
print(df.shape)

print("\nNama Kolom:")
print(df.columns.tolist())

target = "class"

X = df.drop(columns=[target])
y = df[target]

print("\nIndependent Variables:")
print(X.columns.tolist())

print("\nDependent Variable:")
print(target)

print("\nJumlah Independent Variables:", X.shape[1])
print("Jumlah Dependent Variables:", 1)

print("\nJumlah Kategori Target:")
print(y.nunique())

print("\nDistribusi Target:")
print(y.value_counts().sort_index())

print("\nIdentifikasi Klasifikasi:")
if y.nunique() == 2:
    print("Binary Classification")
else:
    print("Bukan Binary Classification")
```

Penjelasan variabel:

1. Variance: Menggambarkan tingkat variasi atau penyebaran nilai piksel hasil transformasi gambar.
2. Skewness: Menggambarkan tingkat kemencengan distribusi nilai.
3. Curtosis: Menggambarkan bentuk distribusi data, terutama karakteristik puncaknya.
4. Entropy: Menggambarkan tingkat kompleksitas atau ketidakaturan pola gambar.
